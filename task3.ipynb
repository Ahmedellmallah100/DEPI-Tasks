```python
def Dimensions (r) :
    Pi = 3.14
    Area = Pi * r * r 
    parameter = 2 * Pi * r
    return (Area , parameter)

Dimensions(5)
```

```python
def convert_celsius_2_fahrenheit (c) :
    f = (c * 9/5) + 32
    return f

convert_celsius_2_fahrenheit(37)
```

```python
def check_season(month):
    if month in ['December', 'January', 'February']:
        return 'Winter'
    elif month in ['March', 'April', 'May']:
        return 'Spring'
    elif month in ['June', 'July', 'August']:
        return 'Summer'
    elif month in ['September', 'October', 'November']:
        return 'Autumn'
    else:
        return 'Invalid month'
    


check_season('April')
```

```python
def calculate_slope(x1, y1, x2, y2):
        slope = (y2 - y1) / (x2 - x1)
        return slope

calculate_slope(2, 3, 5, 7)
```

```python
def solve_quadratic_eqn(a, b, c):
    discriminant = b*b - 4*a*c
    if discriminant > 0:
        root1 = (-b + discriminant**0.5) / (2*a)
        root2 = (-b - discriminant**0.5) / (2*a)
        return (root1, root2)
    elif discriminant == 0:
        root = -b / (2*a)
        return (root,)
    else:
        return "No roots"

    

solve_quadratic_eqn(1, -3, 2)
```

```python
def sum_of_numbers (n) :
    sum = 0
    for i in range (n+1):
        sum += i
    return sum

sum_of_numbers(10)
```

```python
def sum_of_odds (n) :
    sum = 0
    for i in range (n+1):
        if i % 2 != 0:
            sum = sum + i
    return sum

sum_of_odds(30)
```

```python
def sum_of_evens (n) :
    sum = 0
    for i in range (n+1):
        if i % 2 == 0:
            sum = sum + i
    return sum
sum_of_evens(30)
```

```python
def evens_and_odds(n):
    odd=0
    even=0
    for i in range (n+1):
        if i%2==0 :
            even+=1
            
        else:
            odd+=1

    return even ,odd

even,odd=evens_and_odds(100)
print(even)
print(odd)
```
